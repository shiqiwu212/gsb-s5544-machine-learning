2.  Complete the first bullet of the Check-in before Section 7.3.

    Write a function `sqrt_pos_unvec()` that takes in a single value as an argument, and returns the square root if the value is positive. Then, write a for loop that uses this function to construct a new vector where the positive values are square rooted.

```python
import numpy as np
```

```python
def sqrt_pos_unvec(x):
    if x > 0:
        return np.sqrt(x)
    else:
        return x


a_vec = np.array([-2, 1, -3, -9, 7])
new_vec = []

for x in a_vec:
    new_vec.append(sqrt_pos_unvec(x))

new_vec = np.array(new_vec)

new_vec
```

3.  Complete the second bullet of the Check-in before Section 7.3.

    Write a function `sqrt_pos_vec()` that takes in a vector of values, and returns a vector with the positive values square rooted. Do not use a for loop inside your function.

```python
def sqrt_pos_vec(vec):
    result = np.array(vec, dtype=float)
    is_pos = result > 0
    result[is_pos] = np.sqrt(result[is_pos])

    return result


sqrt_pos_vec(a_vec)
```

4.  Complete the Check-in before Section 7.4.
    - a. Write a `sing_verse_3()` function that also lets us change the container (e.g. bottle, can, …) at each step of the loop.
    - b. Use `map()` to sing a few verses.
    - c. What happens if you supply three different drinks, but only two different types of containers?

```python
# a. Write a `sing_verse_3()` function that also lets us change the container (e.g. bottle, can, …) at each step of the loop.
def sing_verse_3(num, drink, container):
    verse = str(num) + " " + container + " of " + drink + " on the wall\n"
    verse = verse + str(num) + " " + container + " of " + drink + "\n"
    verse = verse + "Take one down, pass it around,\n"
    verse = verse + str(num - 1) + " " + container + " of " + drink + " on the wall\n"

    return verse


# b. Use `map()` to sing a few verses.
nums = range(100, 97, -1)
drinks = ["beer", "milk", "lemonade"]
containers = ["bottles", "glasses", "cans"]

song = list(map(sing_verse_3, nums, drinks, containers))

print("".join(song))


# c. What happens if you supply three different drinks, but only two different types of containers?
short_containers = ["bottles", "glasses"]
short_song = list(map(sing_verse_3, nums, drinks, short_containers))

print("Number of verses with two containers:", len(short_song))
print("".join(short_song))
```

Answer: When there are three drinks but only two containers, `map()` produces only two verses because it stops when the shortest iterable runs out. The verses use beer in bottles and milk in glasses. The third drink, lemonade, is not used, and no error occurs.

6. Load the `penguins` dataset from the `palmerpenguins` library, as well as any other libraries you need. Then write a function that takes in information about a penguin, and returns one of the following definitions:

*   "Big Mouth Billy": Male penguins with bill length times bill depth greater than 800.
*   "Dainty Daisy": Female penguins with flipper length less than 5% of body mass.
*   "Average Adelie:" Any Adelie penguins that do not fall into either category.
*   "Other": Any penguins that do not fall into any of the categories.



```python
from palmerpenguins import load_penguins

penguins = load_penguins()

def classify_penguin(penguin):
    if penguin["sex"] == "male" and penguin["bill_length_mm"] * penguin["bill_depth_mm"] > 800:
        return "Big Mouth Billy"
    elif penguin["sex"] == "female" and penguin["flipper_length_mm"] < 0.05 * penguin["body_mass_g"]:
        return "Dainty Daisy"
    elif penguin["species"] == "Adelie":
        return "Average Adelie:"
    else:
        return "Other"

classify_penguin(penguins.iloc[0])
```

7. Use an iterable function to create a new variable in the data frame called `category_name` that adds these labels.

```python
penguins["category_name"] = penguins.apply(classify_penguin, axis=1)

penguins[["species", "sex", "category_name"]].head()
```

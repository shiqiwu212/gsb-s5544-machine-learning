

## Decode a message

Complete the following steps to decode the message.  

1. Remove any spaces before or after each word.
2. Any time you see the word "ugh", with any number of h's, followed by a punctuation mark, delete this.
3. No word should be longer than 16 characters. Drop all extra characters beyond 13 off the end of each word.
4. Replace all instances of exactly 2 a's with exactly 2 e's.
5. Replace all z's with t's.
6. Every word that ends in b, change that to a y.  *Hint: look out for punctuation!*
7. Every word that starts with k, change that to a v.  *Hint: look out for capitalization!*
8. Use `.join()` to recombine all your words into a message.
9. Find the movie this quote is from.

### Step 1: Remove spaces before or after each word.

```python
decoded_message = message.str.strip()

decoded_message
```

### Step 2: Remove "ugh" with repeated h's and the following punctuation mark.

```python
decoded_message = decoded_message.str.replace(r"ugh+[^\w\s]", "", regex=True)

decoded_message
```

### Step 3: Check and shorten long words.

```python
long_words = decoded_message[decoded_message.str.len() > 13]

long_words
```

```python
decoded_message = decoded_message.str.slice(stop=13)

decoded_message.loc[long_words.index]
```

### Step 4: Replace exactly two consecutive a's with two e's.

```python
decoded_message = decoded_message.str.replace(r"(?<!a)aa(?!a)", "ee", regex=True)

decoded_message
```

### Step 5: Replace all z's with t's.

```python
decoded_message = decoded_message.str.replace("z", "t", regex=False)

decoded_message
```

### Step 6: Change a final b to y, keeping any trailing punctuation.

```python
decoded_message = decoded_message.str.replace(r"b(?=[^\w\s]*$)", "y", regex=True)

decoded_message
```

### Step 7: Change an initial k to v, preserving capitalization.

```python
decoded_message = decoded_message.str.replace(r"^k", "v", regex=True)
decoded_message = decoded_message.str.replace(r"^K", "V", regex=True)

decoded_message
```

### Step 8: Join the words into a complete message.

```python
import textwrap

decoded_text = " ".join(decoded_message)

print(textwrap.fill(decoded_text, width=80))
```
